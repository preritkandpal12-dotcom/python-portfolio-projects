```python
import pandas as pd

df = pd.read_csv('weather_by_cities.csv')

max_temp = df['temperature'].max()
print("The maximum temperature recorded was:", max_temp)

ny_weather = df[df['city'] == 'new york']
print("Here is the weather data just for New York:\n", ny_weather)
```

```python
import pandas as pd
df = pd.read_csv('weather_by_cities.csv')
mean_temp = df['temperature'].mean()
print("The mean temp was:", mean_temp)
```

```python
import pandas as pd 
df = pd.read_csv('weather_by_cities.csv')
hottest_days= df.sort_values(by = 'temperature', ascending=False)
print(hottest_days) 


```

```python
import pandas as pd
df = pd.read_csv('weather_by_cities.csv')
double_sort = df.sort_values(by =['city','temperature'])
print(double_sort)
```

```python
import pandas as pd 
df = pd.read_csv('weather_by_cities.csv')
hot_and_rainy = df[(df['temperature'] > 40) & (df['event'] == 'Rain')]
print("Filtered compound data: ")
print(hot_and_rainy) 


```

```python
import pandas as pd
df = pd.read_csv('weather_by_cities.csv')
weather_counts = df['event'].value_counts
print("\nWeather Event Distribution: ")
print(weather_counts)

                 
```
